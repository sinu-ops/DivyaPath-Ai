```python
import pandas as pd
import ast

job_df = pd.read_csv(r"C:\Users\yshel\Desktop\DivyaPath-Ai\data\module2_resume\all_job_post.csv")

# Convert string list to Python list
job_df["job_skill_set"] = job_df["job_skill_set"].apply(
    lambda x: ast.literal_eval(x) if isinstance(x, str) else []
)

# Collect all skills
all_skills = []
for skills in job_df["job_skill_set"]:
    all_skills.extend(skills)

# Normalize
all_skills = [s.strip().lower() for s in all_skills if isinstance(s, str)]

# Unique skills
unique_skills = sorted(set(all_skills))

len(unique_skills), unique_skills[:30]
```

```python
skills_df = pd.DataFrame({"skill": unique_skills})
skills_df.to_csv(
    r"C:\Users\yshel\Desktop\DivyaPath-Ai\data\module2_resume\skills_master.csv",
    index=False
)
```
